```python
import json
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer

# Load small corpus
docs = []
with open("/kaggle/working/wiki_small.jsonl") as f:
    for line in f:
        docs.append(json.loads(line))

print(f"Loaded {len(docs)} docs")

# Build embeddings
encoder = SentenceTransformer("intfloat/e5-base-v2")
texts = ["passage: " + d["contents"] for d in docs]
embeddings = encoder.encode(texts, batch_size=32, show_progress_bar=True, convert_to_numpy=True)

# Build faiss index
faiss.normalize_L2(embeddings)
index = faiss.IndexFlatIP(embeddings.shape[1])
index.add(embeddings)

# Save
faiss.write_index(index, "/kaggle/working/small_e5.index")
print("Done. Index size:", index.ntotal)
```
